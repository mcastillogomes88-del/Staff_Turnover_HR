## **Staff Turnover HR - ETL**

```python
import pandas as pd
import numpy as np
pd.set_option('display.max_columns', None)
```

```python
staff_df = pd.read_csv('Data/staff_turnover_HR.csv')
staff_df.head()  
```

```python
# Informacion general del dataframe
staff_df.describe(include='object')
staff_df.info()
```

```python
# Numero de nulos por columna
staff_df.isnull().sum() 
```

```python
# Selección de columnas relevantes para el análisis
columns_to_keep = ['EmpID','Age','Attrition','BusinessTravel','Department','DistanceFromHome','EducationField','EnvironmentSatisfaction','Gender','JobRole','JobSatisfaction','MaritalStatus','MonthlyIncome','MonthlyRate','NumCompaniesWorked','OverTime','PercentSalaryHike','PerformanceRating','RelationshipSatisfaction','TotalWorkingYears','TrainingTimesLastYear','WorkLifeBalance','YearsAtCompany','YearsInCurrentRole','YearsSinceLastPromotion','YearsWithCurrManager']
staff_df = staff_df[columns_to_keep]
staff_df.shape
```

```python
# Eliminación de duplicados
staff_df.drop_duplicates(inplace=True)
staff_df.shape
```

```python
# Estadisticas descriptivas de las columnas numéricas 
staff_df.describe()
```

```python
# Estadisticas descriptivas de las columnas categóricas
staff_df.describe(include=['object','string'])
```

```python
# Hacemos un conteo de los valores únicos por columna categórica para detectar posibles errores de tipeo o inconsistencias en los datos

# Listamos las columnas que son de tipo object o string
staff_df_Columns_Object = staff_df.select_dtypes(include=['object','string']).columns.tolist()
staff_df_Columns_Object.remove('EmpID')  # Eliminamos la columna 'EmpID' de la lista ya que es un identificador único y no es relevante para este análisis

# Creamos un diccionario extrayendo los valores únicos de cada una
datos_unicos = {col: pd.Series(staff_df[col].unique()) for col in staff_df_Columns_Object}

# Creamos el DataFrame final
df_unicos = pd.DataFrame(datos_unicos)
df_unicos
```

#### ⚠️ **ERRORES DE TIPEO** ⚠️
1. Notamos que en la columna 'BusinessTravel' hay un erros de tipeo ya que 'Travel_Rarely' y 'TravelRarely' son la misma clasificación pero se detectan como diferentes debido al guión bajo.

2. Cambiaremos la clasificación "Non-Travel" por "Non_Travel" para ir acorde al resto.

```python
staff_df.loc[staff_df['BusinessTravel'] == 'TravelRarely', 'BusinessTravel'] = 'Travel_Rarely'
staff_df.loc[staff_df['BusinessTravel'] == 'Non-Travel', 'BusinessTravel'] = 'Non_Travel'
```

```python
print(staff_df['BusinessTravel'].unique())
```

#### ⚠️ **Duplicados en 'EmpID'** ⚠️

Notamos que estos duplicados son generados por la columna 'YearsWithCurrManager', ya que tiene dos valores distintos para un solo EmpID de 3 EmpID. 
Debido a que esta columna tiene valores nulos y reflexionando un poco más al respecto esta no es clave para nuestro análisis por lo que la eliminaremos.

```python
# Verificar si hay duplicados en la columna 'EmpID'
staff_df[staff_df['EmpID'].duplicated(keep=False)]  
```

```python
# Eliminamos la columna 'YearsWithCurrManager' ya que no es relevante para el análisis y puede generar multicolinealidad con otras variables
staff_df = staff_df.drop(columns=['YearsWithCurrManager'])
staff_df.columns # Comprobamos que la columna ha sido eliminada correctamente
```

#### 💡 **Agregación de columnas categoricas**

A continuación haremos la agregación de nuevas columnas para hacer una clasificación en rangos de las columnas 'Age', 'MonthlyIncome', 'YearsAtCompany', 'TotalWorkingYears'. Esto lo haremos en esta parte ya que de esta manera ahorramos calcular estas con DAX de Power BI, con esto buscamos una optimización en el peso del informe generado.

```python

limites = [17, 25, 35, 45, 55, 120]

etiquetas = ["18-25", "26-35", "36-45", "46-55", "55+"]

staff_df["AgeGroup"] = pd.cut(
    staff_df["Age"], bins=limites, labels=etiquetas, right=True
)
```

```python
staff_df['AgeGroup'].value_counts()  # Contamos el número de empleados en cada grupo de edad
```

```python
limites = [1000, 5000, 10000, 15000, 20000]

etiquetas = ["Upto 5k", "5k-10k", "10k-15k", "15k+"]

staff_df["SalarySlab"] = pd.cut(
    staff_df["MonthlyIncome"], bins=limites, labels=etiquetas, right=True
)
```

```python
staff_df['SalarySlab'].value_counts()  # Contamos el número de empleados en cada rango salarial
```

```python
staff_df.to_csv('Data/staff_turnover_HR_cleaned.csv', index=False)  # Guardamos el DataFrame limpio en un nuevo archivo CSV
```
